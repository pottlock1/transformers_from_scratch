```python
if torch.backends.mps.is_available():
    device = torch.device("mps")
elif torch.backends.cuda.is_available():
    device = torch.device("cuda")
else:
    device = torch.device("cpu")
print(f"Selected Device: {device}")
```

```python
torch.manual_seed(42)
batch_size = 4
def get_batch(batch_size = batch_size, data_type = "train"):
    data = train_data if data_type == "train" else val_data
    ix = torch.randint(len(data) - block_size, (batch_size,)) # does not include the max number
    xb = torch.stack([data[i : i + block_size] for i in ix]).to(device)
    yb = torch.stack([data[i + 1 : i + 1 + block_size] for i in ix]).to(device)
    return xb, yb
```

# Model Architecting

```python
@torch.no_grad()
def estimate_loss(model, steps, batch_size = 16):
    losses_dict = {}
    for data_type in ["train", "test"]:
        losses = []
        for step in range(steps):
            x, y = get_batch(batch_size = batch_size, data_type = data_type)
            _, loss = model(x, y)
            losses.append(loss.item())
        losses_dict[data_type] = torch.mean(torch.tensor(losses))
    return losses_dict
```

```python
class SingleHeadAttention(nn.Module):
    def __init__(self, n_emb, head_emb, block_size):
        super().__init__()
        self.query_nn = nn.Linear(n_emb, head_emb)
        self.key_nn = nn.Linear(n_emb, head_emb)
        self.value_nn = nn.Linear(n_emb, head_emb)
        self.register_buffer("tril", torch.tril(torch.ones(block_size, block_size)) == 0)
        
    def forward(self, idx):
        B, T, C = idx.shape
        q = self.query_nn(idx)
        k = self.key_nn(idx)
        v = self.value_nn(idx)
        wei = q @ k.transpose(-2, -1)
        wei = wei.masked_fill(self.tril[:T, :T], -torch.inf)
        print(wei.shape)
        wei = torch.softmax(wei, dim = -1)
        x_sa = wei @ v
        return x_sa
```

```python
class BiGramModel(nn.Module):
    def __init__(self, n_vocab, n_emb, block_size, n_heads):
        super().__init__()
        self.n_vocab = n_vocab
        self.block_size = block_size
        self.n_emb = n_emb
        self.embeddings = nn.Embedding(n_vocab, n_emb) # embedding table of n_vocab x n_emb
        self.pos_embeddings = nn.Embedding(block_size, n_emb)
        self.Head = SingleHeadAttention(n_emb, n_emb//n_heads, block_size)
        self.lm_head = nn.Linear(n_emb, n_vocab) # linear layer from n_emb --> n_vocab

    def forward(self, idx, target = None):
        """
        idx : B x T x C
        target (optional): B x T
        """
        B, T = idx.shape
        # tokens mebedding
        token_emb = self.embeddings(idx)
        # position wise embedding (trainable)
        pos_emb = self.pos_embeddings(torch.arange(start = 0, end = T).to(device))
        # adding positional embedding and token embedding
        x_w_pos = token_emb + pos_emb
        # get the attention weighted embeddings of the input
        x_sa = self.Head(x_w_pos)
        # generating final logits (language model head for next token)
        logits = self.lm_head(x_w_pos)
        
        if target is None:
            loss = None
        else:
            B, T, C = logits.shape
            logits = logits.view(B*T, C)
            target = target.view(B*T)
            loss = F.cross_entropy(logits, target)

        return logits, loss
    @torch.no_grad()
    def generate(self, idx, max_new_tokens = 100):
        """
        generates new tokens untill it reaches max_new_tokens
        idx : B x T x C
        max_new_token : int, how many new tokens needs to be generated
        """
        for _ in tqdm(range(max_new_tokens)):
            # get the logits by passing the idx through model
            idx_block = idx[:,-self.block_size:]
            logits, _ = self(idx) # B x T
            # apply softmax on last token's embeddings
            logits_softmax = F.softmax(logits[:, -1, :], dim = -1)
            # get generation based on the softmaxed probability generations with torch multinomial function
            new_tokens = torch.multinomial(logits_softmax, 1)
            # stack the new tokens with idx
            idx = torch.cat([idx, new_tokens], dim = 1)
        return idx
```

```python
n_vocab = 10000
n_emb = 4
n_heads = 2
block_size = 8
model = BiGramModel(n_vocab, n_emb = n_emb, block_size = block_size, n_heads = n_heads).to(device)
```

## Train the model

```python
log_step = 10
```

```python
opt = torch.optim.AdamW(model.parameters(), lr = 1e-3)
```

```python
losses_dict = estimate_loss(model, 100, batch_size = 8)
losses_dict
```

```python
batch_size = 32
for step in range(1000):
    x, y = get_batch(batch_size=batch_size)
    logits, loss = model(x, y)
    opt.zero_grad(set_to_none=True)
    loss.backward()
    opt.step()
    if step % log_step == 0:
        losses_dict = estimate_loss(model, 100, batch_size = 8)
        print(f"Train loss: {losses_dict['train']:.4f} Val loss: {losses_dict['test']:.4f}")
```
