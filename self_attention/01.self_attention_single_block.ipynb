```python
train_data = data[:int(0.9*len(data))]
val_data = data[int(0.9*len(data)):]
print(f"Len of train data: {len(train_data)}, Len of val data: {len(val_data)}")
```

```python
if torch.backends.mps.is_available():
    device = torch.device("mps")
elif torch.backends.cuda.is_available():
    device = torch.device("cuda")
else:
    device = torch.device("cpu")
print(f"Selected Device: {device}")
```

```python
torch.manual_seed(42)
batch_size = 4
def get_batch(batch_size = batch_size, data_type = "train"):
    data = train_data if data_type == "train" else val_data
    ix = torch.randint(len(data) - block_size, (batch_size,)) # does not include the max number
    xb = torch.stack([data[i : i + block_size] for i in ix]).to(device)
    yb = torch.stack([data[i + 1 : i + 1 + block_size] for i in ix]).to(device)
    return xb, yb
```

```python
xb, yb = get_batch()
```

```python
for row in range(batch_size):
    for i in range(1, block_size):
        print(f"Context is : {xb[row][:i]} Target is : {yb[row][i]}")
```

```python
class BiGramModel(nn.Module):
    def __init__(self, n_vocab):
        super().__init__()
        self.n_vocab = n_vocab
        self.embeddings = nn.Embedding(n_vocab, n_vocab) # embedding table of n_vocab x n_vocab

    def forward(self, idx, target = None):
        """
        idx : B x T x C
        target (optional): B x T
        """
        logits = self.embeddings(idx)
        if target is None:
            loss = None
        else:
            B, T, C = logits.shape
            logits = logits.view(B*T, C)
            target = target.view(B*T)
            loss = F.cross_entropy(logits, target)

        return logits, loss
    
    def generate(self, idx, max_new_tokens = 100):
        """
        generates new tokens untill it reaches max_new_tokens
        idx : B x T x C
        max_new_token : int, how many new tokens needs to be generated
        """
        for _ in tqdm(range(max_new_tokens)):
            # get the logits by passing the idx through model
            logits, _ = self(idx) # B x T
            # apply softmax on last token's embeddings
            logits_softmax = F.softmax(logits[:, -1, :], dim = -1)
            # get generation based on the softmaxed probability generations with torch multinomial function
            new_tokens = torch.multinomial(logits_softmax, 1)
            # stack the new tokens with idx
            idx = torch.cat([idx, new_tokens], dim = 1)
        return idx
```

```python
n_vocab = 10000
model = BiGramModel(n_vocab).to(device)
```

## Train the model

```python
opt = torch.optim.AdamW(model.parameters(), lr = 1e-3)
```

```python
batch_size = 32
for steps in tqdm(range(1000)):
    x, y = get_batch(batch_size=batch_size)
    logits, loss = model(x, y)
    opt.zero_grad(set_to_none=True)
    loss.backward()
    opt.step()
```
